# Gemma 4 31B via OpenRouter

Quick setup to call `google/gemma-4-31b-it` through the OpenRouter API for testing combined stories.

```python
import os

from dotenv import load_dotenv
from openai import OpenAI

load_dotenv()

client = OpenAI(
    base_url="https://openrouter.ai/api/v1",
    api_key=os.environ["OPENROUTER_KEY"],
)

MODEL = "google/gemma-4-31b-it"
```

```python
response = client.chat.completions.create(
    model=MODEL,
    messages=[
        {
            "role": "user",
            "content": "Write a 100-word story simultaneously displaying the following emotions: happy, envious, and conflicted. Do not use these words themselves",
        }
    ],
)
print(response.choices[0].message.content)
```
